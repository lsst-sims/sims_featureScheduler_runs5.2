```python
import numpy as np
import pandas as pd
import matplotlib.pylab as plt
%matplotlib inline
import sqlite3
```

```python
con = sqlite3.connect("templates_cleanup_v5.2.0_0yrs.db")
df = pd.read_sql("select * from observations;", con)
con.close()
```

```python
note_root = []
for note in df["scheduler_note"]:
    split = note.split(",")
    if split[0] == "templates":
        note_root.append(split[0]+split[1])
    else:
        note_root.append(split[0])
```
