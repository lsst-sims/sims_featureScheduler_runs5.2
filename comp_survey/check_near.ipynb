```python
db_file = "comp_survey_v5.2.0_10yrs.db"
run_name = basename(db_file).replace('.db', '')
con = sqlite3.connect(db_file)
df = pd.read_sql('select * from observations;', con)
con.close()
# Convert to a numpy array
visits_array = df.to_records(index=False)
```

```python
near_obs = ["near" in note for note in visits_array["scheduler_note"]]
```

```python
nights = np.unique(visits_array["night"][near_obs])
```

```python
gn = nights[12]
indx = np.where((visits_array["night"] == gn) & (near_obs))[0]
plt.plot(visits_array["azimuth"][indx], visits_array["altitude"][indx], 'ko', alpha=.1)
```
